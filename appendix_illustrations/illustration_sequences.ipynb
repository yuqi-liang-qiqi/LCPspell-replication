## 4. Spell view (selected pairs cited in the text)

Manual spell decomposition for sequences referenced in §2 and the appendix.

````python
def spell_decomposition(states: list[str]) -> list[tuple[str, int]]:
    spells: list[tuple[str, int]] = []
    for state in states:
        if spells and spells[-1][0] == state:
            spells[-1] = (state, spells[-1][1] + 1)
        else:
            spells.append((state, 1))
    return spells


for seq_id in (1, 2, 4, 9, 10):
    letters = TEN_SEQUENCES[seq_id]
    spells = spell_decomposition(letters)
    spell_str = ", ".join(f"({s}, {d})" for s, d in spells)
    print(f"Seq {seq_id:2d}: {', '.join(letters)}  ->  {spell_str}")
````
